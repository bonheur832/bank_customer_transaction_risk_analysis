```python

#  1. Environment Setup


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import warnings
warnings.filterwarnings('ignore')
import os
from datetime import datetime, timedelta
import sys

# Set UTF-8 encoding for Windows compatibility
if sys.platform.startswith('win'):
    os.environ['PYTHONUTF8'] = '1'
    try:
        sys.stdout.reconfigure(encoding='utf-8')
    except AttributeError:
        pass

# Create directories for outputs
os.makedirs('../reports/transaction_analysis', exist_ok=True)
os.makedirs('../reports/figures', exist_ok=True)

# Visualization Settings
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")
plt.rcParams['figure.figsize'] = (12, 8)
plt.rcParams['font.size'] = 12

print(" Environment setup complete")

# ## 2. Data Loading & Initial Exploration

print(" Loading transaction data...")

try:
    # Load the dataset
    df = pd.read_csv('../data/creditcard.csv')
    print(f" Data loaded successfully. Shape: {df.shape}")

    # Display basic info
    print("\n Dataset Overview:")
    print(f"Total records: {df.shape[0]:,}")
    print(f"Total features: {df.shape[1]}")
    print(f"Memory usage: {df.memory_usage().sum() / 1024**2:.1f} MB")

    # Display column information
    print("\n Column Information:")
    print(df.info())

    # Display first few rows
    print("\n First 5 rows:")
    display(df.head())

except FileNotFoundError as e:
    print(f" Error: {e}")
    print("  Please ensure the dataset is located at: ../data/creditcard.csv")
    raise


# 3. Data Quality Check

print(" DATA QUALITY CHECK")
print("=" * 50)

# Check for missing values
print("\n1. Missing Values Analysis:")
missing_data = df.isnull().sum()
missing_percent = (missing_data / len(df)) * 100
missing_df = pd.DataFrame({
    'Missing Values': missing_data,
    'Percentage': missing_percent
})
missing_df = missing_df[missing_df['Missing Values'] > 0]

if len(missing_df) == 0:
    print(" No missing values found!")
else:
    display(missing_df.sort_values('Percentage', ascending=False))

# Check for duplicates
print("\n2. Duplicate Records Analysis:")
duplicates = df.duplicated().sum()
print(f"Duplicate records: {duplicates:,} ({duplicates/len(df)*100:.4f}%)")

if duplicates > 0:
    df = df.drop_duplicates()
    print(f" Removed {duplicates:,} duplicate records")
    print(f"New dataset shape: {df.shape}")

# Check data types
print("\n3. Data Types Analysis:")
print(df.dtypes.value_counts())

# %% [markdown]
# ## 4. Basic Statistics & Fraud Analysis

# %%
print(" BASIC STATISTICS & FRAUD ANALYSIS")
print("=" * 50)

# Calculate basic statistics for Amount
print("\n Transaction Amount Statistics:")
amount_stats = df['Amount'].describe().round(2)
display(amount_stats.to_frame().T)

# Additional statistics
print(f"\n Additional Amount Statistics:")
print(f"Skewness: {df['Amount'].skew():.4f}")
print(f"Kurtosis: {df['Amount'].kurtosis():.4f}")
print(f"95th percentile: ${df['Amount'].quantile(0.95):.2f}")
print(f"99th percentile: ${df['Amount'].quantile(0.99):.2f}")

# Fraud analysis
print("\n Fraud Analysis:")
fraud_counts = df['Class'].value_counts()
fraud_percent = df['Class'].value_counts(normalize=True) * 100

fraud_df = pd.DataFrame({
    'Count': fraud_counts,
    'Percentage': fraud_percent
})
fraud_df.index = ['Legitimate (0)', 'Fraudulent (1)']
display(fraud_df)

# Save fraud statistics
fraud_df.to_csv('../reports/transaction_analysis/fraud_statistics.csv', encoding='utf-8')
print(" Fraud statistics saved")

# %% [markdown]
# ## 5. Feature Analysis (V1-V28)

# %%
print(" FEATURE ANALYSIS (V1-V28)")
print("=" * 50)

# Identify V features
v_features = [col for col in df.columns if col.startswith('V')]
print(f"\nFound {len(v_features)} V-features")

# Analyze V-features
print("\n V-Feature Statistics:")
v_stats_summary = pd.DataFrame({
    'Feature': v_features,
    'Mean': [df[col].mean() for col in v_features],
    'Std': [df[col].std() for col in v_features],
    'Min': [df[col].min() for col in v_features],
    'Max': [df[col].max() for col in v_features],
    'Skewness': [df[col].skew() for col in v_features],
    'Kurtosis': [df[col].kurtosis() for col in v_features]
}).round(4)

# Display top 10 features by variance
v_stats_summary['Variance'] = v_stats_summary['Std'] ** 2
top_var_features = v_stats_summary.sort_values('Variance', ascending=False).head(10)
print("\n Top 10 V-features by Variance:")
display(top_var_features[['Feature', 'Mean', 'Std', 'Variance']])

# Save feature statistics
v_stats_summary.to_csv('../reports/transaction_analysis/v_features_statistics.csv',
                       index=False, encoding='utf-8')
print(" V-feature statistics saved")

# %%
# Correlation analysis between V-features and fraud
print("\n Correlation with Fraud (Class):")
correlations = []
for v_feat in v_features:
    corr = df[v_feat].corr(df['Class'])
    correlations.append({
        'Feature': v_feat,
        'Correlation_with_Fraud': corr,
        'Abs_Correlation': abs(corr)
    })

corr_df = pd.DataFrame(correlations).sort_values('Abs_Correlation', ascending=False)
print("\nTop 10 features most correlated with fraud:")
display(corr_df.head(10).round(4))

# Save correlation analysis
corr_df.to_csv('../reports/transaction_analysis/fraud_correlations.csv',
               index=False, encoding='utf-8')
print(" Fraud correlation analysis saved")

# %% [markdown]
# ## 6. Temporal Analysis (Synthetic Time Data)

# %%
print(" TEMPORAL ANALYSIS")
print("=" * 50)

print("\n  Note: Original dataset doesn't contain timestamps.")
print("Creating synthetic time data for temporal analysis...")

# Generate synthetic transaction dates
np.random.seed(42)
n_transactions = len(df)

# Create synthetic dates (3 months period)
start_date = datetime(2024, 1, 1)
end_date = datetime(2024, 3, 31)

# Generate random dates
date_range = pd.date_range(start=start_date, end=end_date, freq='min')
dates = np.random.choice(date_range, size=n_transactions, replace=True)

# Add to dataframe
df['TransactionDate'] = dates
df['TransactionDate'] = pd.to_datetime(df['TransactionDate'])

# Create time-based features
df['Hour'] = df['TransactionDate'].dt.hour
df['DayOfWeek'] = df['TransactionDate'].dt.dayofweek
df['DayOfMonth'] = df['TransactionDate'].dt.day
df['Month'] = df['TransactionDate'].dt.month
df['WeekOfYear'] = df['TransactionDate'].dt.isocalendar().week

print(f" Synthetic time data created")
print(f"Time period: {df['TransactionDate'].min()} to {df['TransactionDate'].max()}")

# %% [markdown]
# ## 7. Transaction Patterns Analysis

# %%
print(" TRANSACTION PATTERNS ANALYSIS")
print("=" * 50)

# Set TransactionDate as index for time series analysis
df_time = df.set_index('TransactionDate')

# Daily patterns
print("\n Daily Transaction Patterns:")
daily_volume = df_time.resample('D').agg({
    'Amount': ['sum', 'count', 'mean'],
    'Class': 'sum'
})
daily_volume.columns = ['Daily_Amount', 'Daily_Count', 'Avg_Amount', 'Fraud_Count']
daily_volume['Fraud_Rate'] = daily_volume['Fraud_Count'] / daily_volume['Daily_Count']
daily_volume['Day'] = daily_volume.index.strftime('%Y-%m-%d')

print(f"Average daily transactions: {daily_volume['Daily_Count'].mean():.0f}")
print(f"Average daily amount: ${daily_volume['Daily_Amount'].mean():,.0f}")
print(f"Average fraud per day: {daily_volume['Fraud_Count'].mean():.2f}")

# Weekly patterns
print("\n Weekly Patterns:")
df_time['DayOfWeek'] = df_time.index.dayofweek
weekly_pattern = df_time.groupby('DayOfWeek').agg({
    'Amount': ['sum', 'count', 'mean'],
    'Class': 'sum'
})
weekly_pattern.columns = ['Weekly_Amount', 'Weekly_Count', 'Avg_Amount', 'Fraud_Count']
weekly_pattern['Fraud_Rate'] = weekly_pattern['Fraud_Count'] / weekly_pattern['Weekly_Count']
weekly_pattern.index = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']

print("\nWeekly Summary:")
display(weekly_pattern.round(2))

# Hourly patterns
print("\n Hourly Patterns:")
df_time['Hour'] = df_time.index.hour
hourly_pattern = df_time.groupby('Hour').agg({
    'Amount': ['sum', 'count', 'mean'],
    'Class': 'sum'
})
hourly_pattern.columns = ['Hourly_Amount', 'Hourly_Count', 'Avg_Amount', 'Fraud_Count']
hourly_pattern['Fraud_Rate'] = hourly_pattern['Fraud_Count'] / hourly_pattern['Hourly_Count']

print("\nPeak Hours:")
peak_hour_amount = hourly_pattern['Hourly_Amount'].idxmax()
peak_hour_count = hourly_pattern['Hourly_Count'].idxmax()
print(f"Peak hour by amount: {peak_hour_amount}:00 (${hourly_pattern.loc[peak_hour_amount, 'Hourly_Amount']:,.0f})")
print(f"Peak hour by count: {peak_hour_count}:00 ({hourly_pattern.loc[peak_hour_count, 'Hourly_Count']:.0f} transactions)")

# Save patterns
daily_volume.to_csv('../reports/transaction_analysis/daily_patterns.csv', encoding='utf-8')
weekly_pattern.to_csv('../reports/transaction_analysis/weekly_patterns.csv', encoding='utf-8')
hourly_pattern.to_csv('../reports/transaction_analysis/hourly_patterns.csv', encoding='utf-8')

print(" Temporal patterns saved")


# ## 8. Customer ID Generation & Analysis

print(" CUSTOMER ANALYSIS")
print("=" * 50)

print("\n  Note: Original dataset doesn't contain CustomerID.")
print("Generating synthetic customer IDs for analysis...")

# Generate synthetic customer IDs (approximately 50,000 unique customers)
np.random.seed(42)
n_customers = 50000
customer_ids = np.random.choice(range(100000, 100000 + n_customers), size=len(df), replace=True)
df['CustomerID'] = customer_ids

print(f" Generated synthetic CustomerIDs")
print(f"Unique customers: {df['CustomerID'].nunique():,}")
print(f"Average transactions per customer: {len(df) / df['CustomerID'].nunique():.1f}")

# Customer transaction analysis
print("\n Customer Transaction Statistics:")
customer_stats = df.groupby('CustomerID').agg({
    'id': 'count',
    'Amount': ['sum', 'mean', 'std', 'min', 'max'],
    'Class': 'sum'
}).round(2)

customer_stats.columns = ['Transaction_Count', 'Total_Spent', 'Avg_Amount',
                         'Std_Amount', 'Min_Amount', 'Max_Amount', 'Fraud_Count']

# Calculate additional metrics
customer_stats['Fraud_Rate'] = (customer_stats['Fraud_Count'] / customer_stats['Transaction_Count']).round(4)
customer_stats['Customer_Value'] = pd.qcut(customer_stats['Total_Spent'], q=4,
                                          labels=['Low', 'Medium', 'High', 'VIP'])

print(f"\nCustomer Statistics Summary:")
print(f"Average transactions per customer: {customer_stats['Transaction_Count'].mean():.1f}")
print(f"Average spend per customer: ${customer_stats['Total_Spent'].mean():.2f}")
print(f"Median spend per customer: ${customer_stats['Total_Spent'].median():.2f}")
print(f"Customers with fraud: {(customer_stats['Fraud_Count'] > 0).sum():,}")

# Save customer statistics
customer_stats.to_csv('../reports/transaction_analysis/customer_statistics.csv', encoding='utf-8')
print(" Customer statistics saved")


#  9. Visualization Dashboard

print(" GENERATING VISUALIZATION DASHBOARD")
print("=" * 50)

# Create comprehensive visualization dashboard
fig = plt.figure(figsize=(20, 16))

# 1. Transaction Amount Distribution
ax1 = plt.subplot(3, 3, 1)
ax1.hist(df['Amount'], bins=100, edgecolor='black', alpha=0.7, color='skyblue')
ax1.set_xlabel('Transaction Amount ($)')
ax1.set_ylabel('Frequency')
ax1.set_title('Distribution of Transaction Amounts')
ax1.grid(True, alpha=0.3)
ax1.set_xlim(0, df['Amount'].quantile(0.99))  # Remove extreme outliers

# 2. Fraud vs Legitimate Amounts
ax2 = plt.subplot(3, 3, 2)
fraud = df[df['Class'] == 1]['Amount']
legit = df[df['Class'] == 0]['Amount']

ax2.hist(legit, bins=100, alpha=0.5, label='Legitimate', density=True, color='green')
ax2.hist(fraud, bins=30, alpha=0.7, label='Fraudulent', density=True, color='red')
ax2.set_xlabel('Transaction Amount ($)')
ax2.set_ylabel('Density')
ax2.set_title('Transaction Amounts: Fraud vs Legitimate')
ax2.legend()
ax2.grid(True, alpha=0.3)
ax2.set_xlim(0, max(fraud.quantile(0.99), legit.quantile(0.01)))

# 3. Daily Transaction Volume
ax3 = plt.subplot(3, 3, 3)
if len(daily_volume) > 1:
    ax3.plot(daily_volume.index, daily_volume['Daily_Amount']/1000, linewidth=2)
    ax3.set_xlabel('Date')
    ax3.set_ylabel('Amount ($ thousands)')
    ax3.set_title('Daily Transaction Volume')
    ax3.grid(True, alpha=0.3)
    plt.setp(ax3.xaxis.get_majorticklabels(), rotation=45)

# 4. Weekly Pattern
ax4 = plt.subplot(3, 3, 4)
x_pos = np.arange(len(weekly_pattern))
ax4.bar(x_pos, weekly_pattern['Weekly_Count'], alpha=0.7, color='orange')
ax4.set_xlabel('Day of Week')
ax4.set_ylabel('Transaction Count')
ax4.set_title('Weekly Transaction Pattern')
ax4.set_xticks(x_pos)
ax4.set_xticklabels(weekly_pattern.index)
ax4.grid(True, alpha=0.3, axis='y')

# 5. Hourly Pattern
ax5 = plt.subplot(3, 3, 5)
ax5.plot(hourly_pattern.index, hourly_pattern['Hourly_Count'], linewidth=2, color='purple')
ax5.fill_between(hourly_pattern.index, 0, hourly_pattern['Hourly_Count'], alpha=0.3)
ax5.set_xlabel('Hour of Day')
ax5.set_ylabel('Transaction Count')
ax5.set_title('Hourly Transaction Pattern')
ax5.grid(True, alpha=0.3)
ax5.set_xticks(range(0, 24, 3))

# 6. Top V-Features Correlation with Fraud
ax6 = plt.subplot(3, 3, 6)
top_corr = corr_df.head(10).sort_values('Correlation_with_Fraud')
bars = ax6.barh(range(len(top_corr)), top_corr['Correlation_with_Fraud'])
colors = ['red' if x < 0 else 'green' for x in top_corr['Correlation_with_Fraud']]
for i, bar in enumerate(bars):
    bar.set_color(colors[i])
ax6.set_xlabel('Correlation with Fraud')
ax6.set_title('Top Features Correlated with Fraud')
ax6.set_yticks(range(len(top_corr)))
ax6.set_yticklabels(top_corr['Feature'])
ax6.grid(True, alpha=0.3, axis='x')
ax6.axvline(x=0, color='black', linestyle='-', alpha=0.5)

# 7. Customer Spending Distribution
ax7 = plt.subplot(3, 3, 7)
top_customers = customer_stats.nlargest(20, 'Total_Spent')
ax7.barh(range(len(top_customers)), top_customers['Total_Spent']/1000)
ax7.set_xlabel('Total Spent ($ thousands)')
ax7.set_title('Top 20 Customers by Spending')
ax7.set_yticks(range(len(top_customers)))
ax7.set_yticklabels([f'Cust {i}' for i in top_customers.index])
ax7.grid(True, alpha=0.3, axis='x')

# 8. Fraud Rate by Day of Week
ax8 = plt.subplot(3, 3, 8)
ax8.bar(weekly_pattern.index, weekly_pattern['Fraud_Rate']*100, color='salmon', alpha=0.7)
ax8.set_xlabel('Day of Week')
ax8.set_ylabel('Fraud Rate (%)')
ax8.set_title('Fraud Rate by Day of Week')
ax8.grid(True, alpha=0.3, axis='y')
ax8.tick_params(axis='x', rotation=45)

# 9. Amount Distribution by Fraud Status (Box Plot)
ax9 = plt.subplot(3, 3, 9)
box_data = [df[df['Class'] == 0]['Amount'], df[df['Class'] == 1]['Amount']]
bp = ax9.boxplot(box_data, labels=['Legitimate', 'Fraud'], patch_artist=True)
bp['boxes'][0].set_facecolor('lightgreen')
bp['boxes'][1].set_facecolor('lightcoral')
ax9.set_ylabel('Transaction Amount ($)')
ax9.set_title('Amount Distribution by Fraud Status')
ax9.grid(True, alpha=0.3, axis='y')
ax9.set_yscale('log')

plt.tight_layout()
plt.savefig('../reports/figures/transaction_analysis_dashboard.png', dpi=300, bbox_inches='tight')
plt.show()

print(" Visualization dashboard saved")


# 10. Key Performance Indicators (KPIs)


print(" KEY PERFORMANCE INDICATORS")
print("=" * 50)

# Calculate KPIs
total_transactions = len(df)
total_amount = df['Amount'].sum()
avg_transaction = df['Amount'].mean()
fraud_count = df['Class'].sum()
fraud_rate = fraud_count / total_transactions

# Customer KPIs
unique_customers = df['CustomerID'].nunique()
avg_trans_per_customer = total_transactions / unique_customers
avg_spend_per_customer = customer_stats['Total_Spent'].mean()

# Time-based KPIs
peak_day = weekly_pattern['Weekly_Count'].idxmax()
peak_hour = hourly_pattern['Hourly_Count'].idxmax()
avg_daily_transactions = daily_volume['Daily_Count'].mean()

kpis = {
    'KPI': [
        'Total Transactions',
        'Total Transaction Volume',
        'Average Transaction Value',
        'Fraud Count',
        'Fraud Rate',
        'Unique Customers',
        'Average Transactions per Customer',
        'Average Spend per Customer',
        'Peak Transaction Day',
        'Peak Transaction Hour',
        'Average Daily Transactions'
    ],
    'Value': [
        f"{total_transactions:,}",
        f"${total_amount:,.0f}",
        f"${avg_transaction:.2f}",
        f"{fraud_count:,}",
        f"{fraud_rate:.4%}",
        f"{unique_customers:,}",
        f"{avg_trans_per_customer:.1f}",
        f"${avg_spend_per_customer:.2f}",
        f"{peak_day}",
        f"{peak_hour}:00",
        f"{avg_daily_transactions:.0f}"
    ],
    'Target': [
        'Monitor',
        '> $50M',
        '> $100',
        '< 500',
        '< 0.2%',
        '> 40,000',
        '> 10',
        '> $1,000',
        'Weekday',
        'Business Hours',
        '> 5,000'
    ],
    'Status': [
        '📊',
        '✅' if total_amount > 50000000 else '⚠️',
        '✅' if avg_transaction > 100 else '⚠️',
        '✅' if fraud_count < 500 else '⚠️',
        '✅' if fraud_rate < 0.002 else '⚠️',
        '✅' if unique_customers > 40000 else '⚠️',
        '✅' if avg_trans_per_customer > 10 else '⚠️',
        '✅' if avg_spend_per_customer > 1000 else '⚠️',
        '✅' if peak_day in ['Mon', 'Tue', 'Wed', 'Thu', 'Fri'] else '⚠️',
        '✅' if 9 <= peak_hour <= 17 else '⚠️',
        '✅' if avg_daily_transactions > 5000 else '⚠️'
    ]
}

kpi_df = pd.DataFrame(kpis)
print("\n KPI Dashboard:")
display(kpi_df)

# Save KPIs
kpi_df.to_csv('../reports/transaction_analysis/kpi_dashboard.csv', index=False, encoding='utf-8')
print("\n KPI dashboard saved")

# %% [markdown]
# ## 11. Data Preparation for Next Steps

# %%
print(" PREPARING DATA FOR NEXT ANALYSIS STEPS")
print("=" * 50)

# Prepare RFM data for customer segmentation
print("\n🔧 Preparing RFM data...")

# Calculate analysis date (day after last transaction)
analysis_date = df['TransactionDate'].max() + timedelta(days=1)

# Calculate RFM metrics
rfm_data = df.groupby('CustomerID').agg({
    'TransactionDate': lambda x: (analysis_date - x.max()).days,
    'id': 'count',
    'Amount': 'sum'
}).reset_index()

rfm_data.columns = ['CustomerID', 'Recency', 'Frequency', 'Monetary']

print(f" RFM data prepared for {len(rfm_data):,} customers")
print("\n RFM Statistics:")
print(f"  Average Recency: {rfm_data['Recency'].mean():.1f} days")
print(f"  Average Frequency: {rfm_data['Frequency'].mean():.1f} transactions")
print(f"  Average Monetary: ${rfm_data['Monetary'].mean():.2f}")

# Save RFM data
rfm_data.to_csv('../reports/transaction_analysis/rfm_data.csv', index=False, encoding='utf-8')
print(" RFM data saved")

# Save processed transaction data with CustomerID
df.to_csv('../reports/transaction_analysis/processed_transactions.csv', index=False, encoding='utf-8')
print(" Processed transaction data saved")


# 12. Summary Report

print(" GENERATING SUMMARY REPORT")
print("=" * 50)

# Create executive summary
report_date = datetime.now().strftime('%Y-%m-%d %H:%M:%S')
analysis_period = f"{df['TransactionDate'].min().strftime('%Y-%m-%d')} to {df['TransactionDate'].max().strftime('%Y-%m-%d')}"

summary_report = f"""
TRANSACTION ANALYSIS REPORT
===========================
Generated: {report_date}
Analysis Period: {analysis_period}
Total Transactions: {total_transactions:,}

EXECUTIVE SUMMARY
=================

1. TRANSACTION OVERVIEW
-----------------------
• Total Volume: ${total_amount:,.0f}
• Average Transaction: ${avg_transaction:.2f}
• Fraud Rate: {fraud_rate:.4%} ({fraud_count:,} cases)
• Unique Customers: {unique_customers:,}

2. TEMPORAL PATTERNS
--------------------
• Peak Transaction Day: {peak_day}
• Peak Transaction Hour: {peak_hour}:00
• Average Daily Volume: {avg_daily_transactions:.0f} transactions
• Busiest Day Volume: {weekly_pattern['Weekly_Count'].max():.0f} transactions

3. FRAUD CHARACTERISTICS
------------------------
• Fraud Amount Range: ${fraud.min():.2f} - ${fraud.max():.2f}
• Average Fraud Amount: ${fraud.mean():.2f}
• Fraud Detection Rate: Based on V-feature correlations

4. CUSTOMER INSIGHTS
--------------------
• Average Transactions per Customer: {avg_trans_per_customer:.1f}
• Average Spend per Customer: ${avg_spend_per_customer:.2f}
• Top Customer Spend: ${customer_stats['Total_Spent'].max():,.2f}

5. DATA QUALITY
---------------
• Missing Values: None
• Duplicates Removed: {duplicates:,}
• Data Integrity:  Excellent

KEY FINDINGS
============
1. {f'Fraud rate ({fraud_rate:.4%}) is {"above" if fraud_rate > 0.002 else "below"} industry average (0.2%)'}
2. Transaction pattern shows peak activity during {peak_hour}:00 on {peak_day}s
3. {top_corr.iloc[0]['Feature']} has strongest correlation with fraud ({top_corr.iloc[0]['Correlation_with_Fraud']:.4f})
4. Customer base of {unique_customers:,} with healthy transaction frequency

RECOMMENDATIONS
===============
1. Focus fraud detection on transactions with unusual {top_corr.iloc[0]['Feature']} values
2. Increase monitoring during {peak_hour}:00-{(peak_hour+2)%24}:00 hours
3. Develop customer segmentation based on RFM analysis
4. Implement real-time anomaly detection for high-value transactions

NEXT STEPS
==========
1. Run Customer Segmentation analysis (02_customer_segmentation.ipynb)
2. Implement advanced fraud detection algorithms
3. Develop predictive models for fraud prevention

===========================
END OF REPORT
"""

# Save report
report_path = '../reports/transaction_analysis/executive_summary.txt'
with open(report_path, 'w', encoding='utf-8') as f:
    f.write(summary_report)

print(" Summary report generated")
print(f" Report saved to: {report_path}")

# Display report preview
print("\n Report Preview:")
print("=" * 60)
print(summary_report[:500] + "...\n[Full report saved to file]")

# %% [markdown]
# ## 13. Completion

# %%
print(" TRANSACTION ANALYSIS COMPLETE!")
print("=" * 50)
print("\n Generated Files:")
print("   ../reports/transaction_analysis/")
print("     - fraud_statistics.csv")
print("     - v_features_statistics.csv")
print("     - fraud_correlations.csv")
print("     - daily_patterns.csv")
print("     - weekly_patterns.csv")
print("     - hourly_patterns.csv")
print("     - customer_statistics.csv")
print("     - kpi_dashboard.csv")
print("     - rfm_data.csv")
print("     - processed_transactions.csv")
print("     - executive_summary.txt")
print("\n   ../reports/figures/")
print("     - transaction_analysis_dashboard.png")
print("\n  Next Step: Run 02_customer_segmentation.ipynb")
```
