# Bank Transaction Risk Analysis
###  Part 3: Fraud & Anomaly Detection
### Dataset: 568,630 transactions with V1-V28 features
### Objective: Detect fraudulent transactions using statistical and ML methods




```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler, RobustScaler
from sklearn.metrics import (classification_report, confusion_matrix,
                           roc_curve, auc, precision_recall_curve,
                           precision_score, recall_score, f1_score)
from scipy import stats
import warnings
warnings.filterwarnings('ignore')
import os
from datetime import datetime
import json
import joblib

# Create output directories
os.makedirs('../reports/anomaly_detection', exist_ok=True)
os.makedirs('../reports/figures', exist_ok=True)
os.makedirs('../models', exist_ok=True)

# Visualization settings
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")
plt.rcParams['figure.figsize'] = (12, 8)
plt.rcParams['font.size'] = 12

print("Environment setup complete")

# %% [markdown]
# ## 2. Load & Prepare Data

# %%
print("Loading data for anomaly detection...")

try:
    # Try to load enriched transaction data
    df = pd.read_csv('../reports/customer_segmentation/enriched_transactions.csv')
    print(f"Enriched transaction data loaded: {len(df):,} records")

    # Check for necessary columns
    required_columns = ['Class', 'Amount'] + [f'V{i}' for i in range(1, 29)]
    missing_cols = [col for col in required_columns if col not in df.columns]

    if missing_cols:
        print(f"Missing columns: {missing_cols}")
        print("Loading original data instead...")
        df = pd.read_csv('../data/creditcard.csv')
    else:
        print("All required columns present")

except FileNotFoundError:
    print("Enriched data not found. Loading original data...")
    df = pd.read_csv('../data/creditcard.csv')
    print(f"Original data loaded: {len(df):,} records")

# Display basic info
print(f"\nDataset Overview:")
print(f"Shape: {df.shape}")
print(f"Features: {df.columns.tolist()[:10]}...")

print(f"\nFraud Distribution:")
fraud_counts = df['Class'].value_counts()
fraud_percent = df['Class'].value_counts(normalize=True) * 100
fraud_df = pd.DataFrame({'Count': fraud_counts, 'Percentage': fraud_percent})
fraud_df.index = ['Legitimate (0)', 'Fraudulent (1)']
display(fraud_df)

# Save fraud statistics
fraud_df.to_csv('../reports/anomaly_detection/fraud_distribution.csv', encoding='utf-8')
print("Fraud distribution saved")

# %% [markdown]
# ## 3. Feature Analysis & Selection

# %%
print("FEATURE ANALYSIS & SELECTION")
print("=" * 50)

# Identify V features
v_features = [col for col in df.columns if col.startswith('V')]
print(f"\nFound {len(v_features)} V-features")

# Analyze feature importance for fraud detection
print("\nFeature Correlation with Fraud:")
correlations = []
for feature in v_features + ['Amount']:
    if feature in df.columns:
        corr = df[feature].corr(df['Class'])
        correlations.append({
            'Feature': feature,
            'Correlation': corr,
            'Abs_Correlation': abs(corr)
        })

corr_df = pd.DataFrame(correlations).sort_values('Abs_Correlation', ascending=False)
print("\nTop 15 features most correlated with fraud:")
display(corr_df.head(15).round(4))

# Select top features for anomaly detection
top_n_features = 15
selected_features = corr_df.head(top_n_features)['Feature'].tolist()
print(f"\nSelected {len(selected_features)} features for anomaly detection:")
print(selected_features)

# Save feature selection
corr_df.to_csv('../reports/anomaly_detection/feature_correlations.csv', index=False, encoding='utf-8')
pd.DataFrame({'Selected_Features': selected_features}).to_csv(
    '../reports/anomaly_detection/selected_features.csv', index=False, encoding='utf-8')

print("Feature analysis saved")

# %% [markdown]
# ## 4. Statistical Anomaly Detection (Z-Score Method)

# %%
print("STATISTICAL ANOMALY DETECTION (Z-SCORE)")
print("=" * 50)

print("\nCalculating statistical anomalies...")

# Select features for statistical analysis
stat_features = ['Amount'] + [f'V{i}' for i in [1, 2, 3, 4, 7, 9, 10, 11, 12, 14, 16, 17, 18]]
stat_features = [f for f in stat_features if f in df.columns]

# Calculate Z-scores
z_scores = pd.DataFrame()
for feature in stat_features:
    # Handle infinite/nan values
    data_clean = df[feature].replace([np.inf, -np.inf], np.nan).fillna(df[feature].median())
    z_scores[feature] = np.abs(stats.zscore(data_clean))

# Calculate overall anomaly score
z_scores['Overall_ZScore'] = z_scores.max(axis=1)
df['ZScore'] = z_scores['Overall_ZScore']

# Set multiple thresholds for analysis
thresholds = [2.5, 3, 3.5, 4]
zscore_results = []

for threshold in thresholds:
    df[f'ZScore_Anomaly_{threshold}'] = df['ZScore'] > threshold
    anomalies = df[f'ZScore_Anomaly_{threshold}'].sum()
    anomaly_rate = anomalies / len(df)

    # Calculate performance metrics
    if 'Class' in df.columns:
        tp = ((df[f'ZScore_Anomaly_{threshold}'] == True) & (df['Class'] == 1)).sum()
        fp = ((df[f'ZScore_Anomaly_{threshold}'] == True) & (df['Class'] == 0)).sum()
        fn = ((df[f'ZScore_Anomaly_{threshold}'] == False) & (df['Class'] == 1)).sum()
        tn = ((df[f'ZScore_Anomaly_{threshold}'] == False) & (df['Class'] == 0)).sum()

        precision = tp / (tp + fp) if (tp + fp) > 0 else 0
        recall = tp / (tp + fn) if (tp + fn) > 0 else 0
        f1 = 2 * precision * recall / (precision + recall) if (precision + recall) > 0 else 0
        accuracy = (tp + tn) / (tp + tn + fp + fn)
    else:
        precision = recall = f1 = accuracy = np.nan

    zscore_results.append({
        'Threshold': threshold,
        'Anomalies_Detected': anomalies,
        'Anomaly_Rate': anomaly_rate,
        'Precision': precision,
        'Recall': recall,
        'F1_Score': f1,
        'Accuracy': accuracy
    })

zscore_metrics_df = pd.DataFrame(zscore_results)
print("\nZ-Score Method Performance at Different Thresholds:")
display(zscore_metrics_df.round(4))

# Select optimal threshold based on F1 score
if 'Class' in df.columns:
    optimal_threshold_idx = zscore_metrics_df['F1_Score'].idxmax()
    optimal_threshold = zscore_metrics_df.loc[optimal_threshold_idx, 'Threshold']
    df['ZScore_Anomaly'] = df[f'ZScore_Anomaly_{optimal_threshold}']
    print(f"\nOptimal Z-score threshold: {optimal_threshold} (F1: {zscore_metrics_df.loc[optimal_threshold_idx, 'F1_Score']:.4f})")
else:
    optimal_threshold = 3  # Default
    df['ZScore_Anomaly'] = df['ZScore'] > optimal_threshold

print(f"\nZ-Score Analysis Summary:")
print(f"   Optimal threshold: {optimal_threshold}")
print(f"   Anomalies detected: {df['ZScore_Anomaly'].sum():,}")
print(f"   Anomaly rate: {df['ZScore_Anomaly'].mean():.4%}")
print(f"   Max Z-score: {df['ZScore'].max():.2f}")

# Save Z-score results
zscore_metrics_df.to_csv('../reports/anomaly_detection/zscore_metrics.csv', index=False, encoding='utf-8')
print("Z-score analysis saved")

# %% [markdown]
# ## 5. Isolation Forest Anomaly Detection

# %%
print("ISOLATION FOREST ANOMALY DETECTION")
print("=" * 50)

print("\nPreparing data for Isolation Forest...")

# Prepare features
iso_features = selected_features.copy()
if 'Class' in iso_features:
    iso_features.remove('Class')
if 'ZScore' in iso_features:
    iso_features.remove('ZScore')

X_iso = df[iso_features].copy()

# Handle missing/infinite values
X_iso = X_iso.replace([np.inf, -np.inf], np.nan)
for col in X_iso.columns:
    X_iso[col] = X_iso[col].fillna(X_iso[col].median())

# Scale features
scaler = RobustScaler()
X_scaled = scaler.fit_transform(X_iso)
X_scaled_df = pd.DataFrame(X_scaled, columns=iso_features)

print(f"Prepared {X_scaled_df.shape[1]} features for Isolation Forest")

# %%
print("\nTraining Isolation Forest models...")

# Test different contamination parameters
contaminations = [0.001, 0.002, 0.005, 0.01, 0.02]
iso_results = []

for contamination in contaminations:
    print(f"  Testing contamination={contamination}...")

    iso_forest = IsolationForest(
        n_estimators=100,
        contamination=contamination,
        random_state=42,
        n_jobs=-1,
        verbose=0
    )

    # Fit and predict
    iso_predictions = iso_forest.fit_predict(X_scaled_df)

    # Convert predictions: -1 for anomalies, 1 for normal
    df[f'Isolation_Anomaly_{contamination}'] = iso_predictions == -1
    df[f'Isolation_Score_{contamination}'] = -iso_forest.decision_function(X_scaled_df)

    anomalies = df[f'Isolation_Anomaly_{contamination}'].sum()
    anomaly_rate = anomalies / len(df)

    # Calculate performance metrics
    if 'Class' in df.columns:
        tp = ((df[f'Isolation_Anomaly_{contamination}'] == True) & (df['Class'] == 1)).sum()
        fp = ((df[f'Isolation_Anomaly_{contamination}'] == True) & (df['Class'] == 0)).sum()
        fn = ((df[f'Isolation_Anomaly_{contamination}'] == False) & (df['Class'] == 1)).sum()
        tn = ((df[f'Isolation_Anomaly_{contamination}'] == False) & (df['Class'] == 0)).sum()

        precision = tp / (tp + fp) if (tp + fp) > 0 else 0
        recall = tp / (tp + fn) if (tp + fn) > 0 else 0
        f1 = 2 * precision * recall / (precision + recall) if (precision + recall) > 0 else 0
        accuracy = (tp + tn) / (tp + tn + fp + fn)
    else:
        precision = recall = f1 = accuracy = np.nan

    iso_results.append({
        'Contamination': contamination,
        'Anomalies_Detected': anomalies,
        'Anomaly_Rate': anomaly_rate,
        'Precision': precision,
        'Recall': recall,
        'F1_Score': f1,
        'Accuracy': accuracy
    })

iso_metrics_df = pd.DataFrame(iso_results)
print("\nIsolation Forest Performance at Different Contaminations:")
display(iso_metrics_df.round(4))

# Select optimal contamination
if 'Class' in df.columns:
    optimal_contamination_idx = iso_metrics_df['F1_Score'].idxmax()
    optimal_contamination = iso_metrics_df.loc[optimal_contamination_idx, 'Contamination']
else:
    # Use expected fraud rate if Class not available
    optimal_contamination = 0.005  # 0.5%
    optimal_contamination_idx = iso_metrics_df['Contamination'].sub(optimal_contamination).abs().idxmin()

print(f"\nOptimal contamination: {optimal_contamination}")

# Use optimal model
df['Isolation_Anomaly'] = df[f'Isolation_Anomaly_{optimal_contamination}']
df['Isolation_Score'] = df[f'Isolation_Score_{optimal_contamination}']

print(f"\nIsolation Forest Summary:")
print(f"   Optimal contamination: {optimal_contamination}")
print(f"   Anomalies detected: {df['Isolation_Anomaly'].sum():,}")
print(f"   Anomaly rate: {df['Isolation_Anomaly'].mean():.4%}")

# Save the best model
optimal_iso = IsolationForest(
    n_estimators=100,
    contamination=optimal_contamination,
    random_state=42,
    n_jobs=-1
)
optimal_iso.fit(X_scaled_df)
joblib.dump(optimal_iso, '../models/isolation_forest_model.pkl')
joblib.dump(scaler, '../models/robust_scaler.pkl')

print("Isolation Forest model saved")

# Save metrics
iso_metrics_df.to_csv('../reports/anomaly_detection/isolation_forest_metrics.csv', index=False, encoding='utf-8')
print("Isolation Forest metrics saved")

# %% [markdown]
# ## 6. Combined Detection Approach

# %%
print("COMBINED ANOMALY DETECTION APPROACH")
print("=" * 50)

print("\nCombining detection methods...")

# Normalize scores for combination
df['ZScore_Normalized'] = (df['ZScore'] - df['ZScore'].min()) / (df['ZScore'].max() - df['ZScore'].min())
df['Isolation_Score_Normalized'] = (df['Isolation_Score'] - df['Isolation_Score'].min()) / (df['Isolation_Score'].max() - df['Isolation_Score'].min())

# Method 1: Weighted combination
weights = {'zscore': 0.4, 'isolation': 0.6}
df['Combined_Score'] = (
    weights['zscore'] * df['ZScore_Normalized'] +
    weights['isolation'] * df['Isolation_Score_Normalized']
)

# Test different thresholds for combined score
combined_thresholds = np.arange(0.1, 0.9, 0.1)
combined_results = []

for threshold in combined_thresholds:
    df[f'Combined_Anomaly_{threshold:.1f}'] = df['Combined_Score'] > threshold
    anomalies = df[f'Combined_Anomaly_{threshold:.1f}'].sum()
    anomaly_rate = anomalies / len(df)

    if 'Class' in df.columns:
        tp = ((df[f'Combined_Anomaly_{threshold:.1f}'] == True) & (df['Class'] == 1)).sum()
        fp = ((df[f'Combined_Anomaly_{threshold:.1f}'] == True) & (df['Class'] == 0)).sum()
        fn = ((df[f'Combined_Anomaly_{threshold:.1f}'] == False) & (df['Class'] == 1)).sum()
        tn = ((df[f'Combined_Anomaly_{threshold:.1f}'] == False) & (df['Class'] == 0)).sum()

        precision = tp / (tp + fp) if (tp + fp) > 0 else 0
        recall = tp / (tp + fn) if (tp + fn) > 0 else 0
        f1 = 2 * precision * recall / (precision + recall) if (precision + recall) > 0 else 0
        accuracy = (tp + tn) / (tp + tn + fp + fn)
    else:
        precision = recall = f1 = accuracy = np.nan

    combined_results.append({
        'Threshold': threshold,
        'Anomalies_Detected': anomalies,
        'Anomaly_Rate': anomaly_rate,
        'Precision': precision,
        'Recall': recall,
        'F1_Score': f1,
        'Accuracy': accuracy
    })

combined_metrics_df = pd.DataFrame(combined_results)
print("\nCombined Method Performance at Different Thresholds:")
display(combined_metrics_df.round(4))

# Select optimal threshold
if 'Class' in df.columns:
    optimal_combined_idx = combined_metrics_df['F1_Score'].idxmax()
    optimal_combined_threshold = combined_metrics_df.loc[optimal_combined_idx, 'Threshold']
else:
    # Set threshold to detect expected fraud rate
    expected_fraud_rate = 0.002  # 0.2%
    combined_metrics_df['Rate_Diff'] = abs(combined_metrics_df['Anomaly_Rate'] - expected_fraud_rate)
    optimal_combined_idx = combined_metrics_df['Rate_Diff'].idxmin()
    optimal_combined_threshold = combined_metrics_df.loc[optimal_combined_idx, 'Threshold']

print(f"\nOptimal combined threshold: {optimal_combined_threshold:.2f}")

# Set final anomaly flag
df['Final_Anomaly'] = df['Combined_Score'] > optimal_combined_threshold

print(f"\nCombined Method Summary:")
print(f"   Optimal threshold: {optimal_combined_threshold:.2f}")
print(f"   Anomalies detected: {df['Final_Anomaly'].sum():,}")
print(f"   Anomaly rate: {df['Final_Anomaly'].mean():.4%}")

# Save combined metrics
combined_metrics_df.to_csv('../reports/anomaly_detection/combined_metrics.csv', index=False, encoding='utf-8')
print("Combined method metrics saved")

# %% [markdown]
# ## 7. Performance Evaluation

# %%
print("PERFORMANCE EVALUATION")
print("=" * 50)

if 'Class' in df.columns:
    print("\nEvaluating detection performance...")

    # Compare all methods
    methods = ['ZScore_Anomaly', 'Isolation_Anomaly', 'Final_Anomaly']
    method_names = ['Z-Score', 'Isolation Forest', 'Combined Method']

    evaluation_results = []

    for method, name in zip(methods, method_names):
        # Calculate confusion matrix
        y_true = df['Class']
        y_pred = df[method]

        # Calculate metrics
        precision = precision_score(y_true, y_pred, zero_division=0)
        recall = recall_score(y_true, y_pred, zero_division=0)
        f1 = f1_score(y_true, y_pred, zero_division=0)
        accuracy = (y_true == y_pred).mean()

        # Calculate confusion matrix values
        tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()

        # Calculate additional metrics
        specificity = tn / (tn + fp) if (tn + fp) > 0 else 0
        false_positive_rate = fp / (fp + tn) if (fp + tn) > 0 else 0

        evaluation_results.append({
            'Method': name,
            'True_Positives': tp,
            'False_Positives': fp,
            'False_Negatives': fn,
            'True_Negatives': tn,
            'Precision': precision,
            'Recall (Sensitivity)': recall,
            'Specificity': specificity,
            'F1_Score': f1,
            'Accuracy': accuracy,
            'False_Positive_Rate': false_positive_rate
        })

    performance_df = pd.DataFrame(evaluation_results)
    print("\nPerformance Comparison of All Methods:")
    display(performance_df.round(4))

    # Save performance metrics
    performance_df.to_csv('../reports/anomaly_detection/performance_comparison.csv', index=False, encoding='utf-8')
    print("Performance comparison saved")

    # Generate classification reports
    print("\nClassification Report - Combined Method:")
    print(classification_report(df['Class'], df['Final_Anomaly'],
                                target_names=['Legitimate', 'Fraudulent']))

    # Save classification report
    report_dict = classification_report(df['Class'], df['Final_Anomaly'],
                                        target_names=['Legitimate', 'Fraudulent'],
                                        output_dict=True)
    report_df = pd.DataFrame(report_dict).transpose()
    report_df.to_csv('../reports/anomaly_detection/classification_report.csv', encoding='utf-8')
    print("Classification report saved")

# %% [markdown]
# ## 8. Risk Scoring & Categorization

# %%
print("RISK SCORING & CATEGORIZATION")
print("=" * 50)

print("\nCalculating risk scores and categories...")

# Calculate comprehensive risk score
df['Risk_Score'] = (
    0.3 * df['ZScore_Normalized'] +
    0.4 * df['Isolation_Score_Normalized'] +
    0.2 * (df['Amount'] / df['Amount'].quantile(0.99)).clip(0, 1) +
    0.1 * df['Combined_Score']
)

# Categorize risk levels
def categorize_risk(score):
    if score >= 0.8:
        return 'Critical'
    elif score >= 0.6:
        return 'High'
    elif score >= 0.4:
        return 'Medium'
    elif score >= 0.2:
        return 'Low'
    else:
        return 'Very Low'

df['Risk_Level'] = df['Risk_Score'].apply(categorize_risk)

# Analyze risk distribution
risk_distribution = df['Risk_Level'].value_counts().sort_index()
risk_percentage = risk_distribution / len(df) * 100

risk_analysis = pd.DataFrame({
    'Transaction_Count': risk_distribution,
    'Percentage': risk_percentage.round(2)
})

print("\nRisk Level Distribution:")
display(risk_analysis)

# Calculate average amounts by risk level
risk_amounts = df.groupby('Risk_Level').agg({
    'Amount': ['mean', 'sum', 'count'],
    'Final_Anomaly': 'mean',
    'Class': 'mean' if 'Class' in df.columns else None
}).round(2)

risk_amounts.columns = ['Avg_Amount', 'Total_Amount', 'Count', 'Anomaly_Rate', 'Actual_Fraud_Rate' if 'Class' in df.columns else None]
risk_amounts = risk_amounts.dropna(axis=1, how='all')

print("\nRisk Level Financial Analysis:")
display(risk_amounts)

# Identify high-risk transactions
high_risk_transactions = df[df['Risk_Level'].isin(['Critical', 'High'])].copy()
print(f"\nHigh-Risk Transactions Identified: {len(high_risk_transactions):,}")
print(f"   Critical Risk: {(df['Risk_Level'] == 'Critical').sum():,}")
print(f"   High Risk: {(df['Risk_Level'] == 'High').sum():,}")

# Handle the amount safely
if 'high_risk_transactions' in locals() and len(high_risk_transactions) > 0:
    total_high_risk_amount = high_risk_transactions['Amount'].sum()
    print(f"   Total High-Risk Amount: ${total_high_risk_amount:,.2f}")
else:
    print(f"   Total High-Risk Amount: $0.00")

# Save risk analysis
risk_analysis.to_csv('../reports/anomaly_detection/risk_level_distribution.csv', encoding='utf-8')
risk_amounts.to_csv('../reports/anomaly_detection/risk_level_financials.csv', encoding='utf-8')

if 'high_risk_transactions' in locals() and len(high_risk_transactions) > 0:
    high_risk_transactions.to_csv('../reports/anomaly_detection/high_risk_transactions.csv', index=False, encoding='utf-8')

print("Risk analysis saved")


# 9. Segment-Based Risk Analysis

print("SEGMENT-BASED RISK ANALYSIS")
print("=" * 50)

# Check if segment data is available
segment_columns = ['RFM_Segment', 'Cluster', 'Cluster_Name']
available_segments = [col for col in segment_columns if col in df.columns]

if available_segments:
    print(f"\nAnalyzing risk by segments: {available_segments}")

    for segment_col in available_segments:
        print(f"\nRisk Analysis by {segment_col}:")

        segment_risk = df.groupby(segment_col).agg({
            'id': 'count',
            'Amount': 'mean',
            'Risk_Score': 'mean',
            'Final_Anomaly': 'mean',
            'Class': 'mean' if 'Class' in df.columns else None
        }).round(4)

        segment_risk = segment_risk.rename(columns={'id': 'Transaction_Count'})
        segment_risk['Anomaly_Rate'] = segment_risk['Final_Anomaly']
        segment_risk['Pct_of_Transactions'] = (segment_risk['Transaction_Count'] / len(df) * 100).round(2)

        if 'Class' in df.columns:
            segment_risk['Actual_Fraud_Rate'] = segment_risk['Class']

        # Sort by risk score
        segment_risk = segment_risk.sort_values('Risk_Score', ascending=False)

        print(f"\nTop 5 segments by risk score:")
        display(segment_risk.head())

        # Save segment risk analysis
        segment_risk.to_csv(f'../reports/anomaly_detection/risk_by_{segment_col.lower()}.csv', encoding='utf-8')

    print("Segment risk analysis saved")
else:
    print("No segment data available for analysis")


# 10. Feature Importance Analysis

print("FEATURE IMPORTANCE ANALYSIS")
print("=" * 50)

print("\nAnalyzing feature importance for fraud detection...")

# Calculate feature importance using multiple methods
feature_importance = pd.DataFrame({'Feature': iso_features})

# Method 1: Correlation with fraud
if 'Class' in df.columns:
    correlations = []
    for feature in iso_features:
        corr = df[feature].corr(df['Class'])
        correlations.append(abs(corr))
    feature_importance['Correlation_Importance'] = correlations

# Method 2: Variance in anomalous vs normal transactions
if 'Final_Anomaly' in df.columns:
    variance_ratios = []
    for feature in iso_features:
        normal_var = df[df['Final_Anomaly'] == False][feature].var()
        anomaly_var = df[df['Final_Anomaly'] == True][feature].var()
        ratio = anomaly_var / normal_var if normal_var > 0 else 0
        variance_ratios.append(ratio)
    feature_importance['Variance_Ratio'] = variance_ratios

# Method 3: Mean difference between anomalous and normal
if 'Final_Anomaly' in df.columns:
    mean_diffs = []
    for feature in iso_features:
        normal_mean = df[df['Final_Anomaly'] == False][feature].mean()
        anomaly_mean = df[df['Final_Anomaly'] == True][feature].mean()
        mean_diffs.append(abs(anomaly_mean - normal_mean))
    feature_importance['Mean_Difference'] = mean_diffs

# Normalize and combine importance scores
for col in feature_importance.columns[1:]:
    if col in feature_importance.columns:
        min_val = feature_importance[col].min()
        max_val = feature_importance[col].max()
        if max_val > min_val:
            feature_importance[f'{col}_Normalized'] = (feature_importance[col] - min_val) / (max_val - min_val)
        else:
            feature_importance[f'{col}_Normalized'] = 0

# Calculate combined importance
normalized_cols = [col for col in feature_importance.columns if 'Normalized' in col]
if normalized_cols:
    feature_importance['Combined_Importance'] = feature_importance[normalized_cols].mean(axis=1)
    feature_importance = feature_importance.sort_values('Combined_Importance', ascending=False)

    print("\nTop 10 Most Important Features for Fraud Detection:")
    display(feature_importance.head(10).round(4))

    # Save feature importance
    feature_importance.to_csv('../reports/anomaly_detection/feature_importance.csv', index=False, encoding='utf-8')
    print("Feature importance analysis saved")

# %% [markdown]
# ## 11. Visualization Dashboard

# %%
print("GENERATING ANOMALY DETECTION VISUALIZATIONS")
print("=" * 50)

# Create comprehensive visualization dashboard
fig = plt.figure(figsize=(20, 16))

# 1. Fraud Distribution
ax1 = plt.subplot(3, 3, 1)
if 'Class' in df.columns:
    fraud_labels = ['Legitimate', 'Fraudulent']
    fraud_values = [fraud_counts[0], fraud_counts[1]]
    colors = ['lightgreen', 'salmon']
    ax1.pie(fraud_values, labels=fraud_labels, autopct='%1.3f%%', colors=colors, startangle=90)
    ax1.set_title('Fraud Distribution (Highly Imbalanced)')

# 2. Risk Level Distribution
ax2 = plt.subplot(3, 3, 2)
risk_colors = {'Very Low': 'green', 'Low': 'lightgreen', 'Medium': 'yellow',
               'High': 'orange', 'Critical': 'red'}
risk_colors_list = [risk_colors[level] for level in risk_analysis.index]
bars = ax2.bar(risk_analysis.index, risk_analysis['Percentage'], color=risk_colors_list)
ax2.set_xlabel('Risk Level')
ax2.set_ylabel('Percentage (%)')
ax2.set_title('Transaction Risk Level Distribution')
ax2.tick_params(axis='x', rotation=45)
ax2.grid(True, alpha=0.3, axis='y')

# Add value labels
for bar in bars:
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height + 0.5,
            f'{height:.1f}%', ha='center', va='bottom')

# 3. Combined Score Distribution
ax3 = plt.subplot(3, 3, 3)
ax3.hist(df['Combined_Score'].fillna(0), bins=50, edgecolor='black', alpha=0.7, color='skyblue')
ax3.axvline(x=optimal_combined_threshold, color='red', linestyle='--',
            label=f'Threshold ({optimal_combined_threshold:.2f})')
ax3.set_xlabel('Combined Anomaly Score')
ax3.set_ylabel('Frequency')
ax3.set_title('Distribution of Combined Anomaly Scores')
ax3.legend()
ax3.grid(True, alpha=0.3)

# 4. Method Performance Comparison
ax4 = plt.subplot(3, 3, 4)
if 'Class' in df.columns and 'performance_df' in locals():
    methods = performance_df['Method']
    precision = performance_df['Precision'].values
    recall = performance_df['Recall (Sensitivity)'].values

    x = np.arange(len(methods))
    width = 0.35

    bars1 = ax4.bar(x - width/2, precision, width, label='Precision', color='blue', alpha=0.7)
    bars2 = ax4.bar(x + width/2, recall, width, label='Recall', color='green', alpha=0.7)

    ax4.set_xlabel('Detection Method')
    ax4.set_ylabel('Score')
    ax4.set_title('Detection Method Performance')
    ax4.set_xticks(x)
    ax4.set_xticklabels(methods, rotation=45, ha='right')
    ax4.legend()
    ax4.grid(True, alpha=0.3, axis='y')
    ax4.set_ylim(0, 1)

# 5. Feature Importance
ax5 = plt.subplot(3, 3, 5)
if 'feature_importance' in locals() and 'Combined_Importance' in feature_importance.columns:
    top_features = feature_importance.head(10).sort_values('Combined_Importance', ascending=True)
    bars = ax5.barh(range(len(top_features)), top_features['Combined_Importance'].values)
    ax5.set_xlabel('Importance Score')
    ax5.set_title('Top 10 Features for Fraud Detection')
    ax5.set_yticks(range(len(top_features)))
    ax5.set_yticklabels(top_features['Feature'].tolist())
    ax5.grid(True, alpha=0.3, axis='x')

    # Add value labels
    for i, bar in enumerate(bars):
        width = bar.get_width()
        ax5.text(width + 0.01, bar.get_y() + bar.get_height()/2,
                f'{width:.3f}', ha='left', va='center')

# 6. Amount vs Risk Score
ax6 = plt.subplot(3, 3, 6)
sample_size = min(5000, len(df))
sample_indices = np.random.choice(len(df), sample_size, replace=False)
sample_data = df.iloc[sample_indices]

scatter = ax6.scatter(sample_data['Amount'], sample_data['Risk_Score'],
                     c=sample_data['Risk_Score'], cmap='YlOrRd', alpha=0.6, s=20)
ax6.set_xlabel('Transaction Amount ($)')
ax6.set_ylabel('Risk Score')
ax6.set_title('Transaction Amount vs Risk Score (Sampled)')
ax6.set_xscale('log')
ax6.grid(True, alpha=0.3)

cbar = plt.colorbar(scatter, ax=ax6)
cbar.set_label('Risk Score')

# 7. Confusion Matrix
ax7 = plt.subplot(3, 3, 7)
if 'Class' in df.columns and 'Final_Anomaly' in df.columns:
    cm = confusion_matrix(df['Class'], df['Final_Anomaly'])

    im = ax7.imshow(cm, interpolation='nearest', cmap=plt.cm.Blues)
    ax7.set_title('Confusion Matrix (Combined Method)')

    # Add text annotations
    thresh = cm.max() / 2.
    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            ax7.text(j, i, format(cm[i, j], 'd'),
                    ha="center", va="center",
                    color="white" if cm[i, j] > thresh else "black")

    ax7.set_ylabel('Actual')
    ax7.set_xlabel('Predicted')
    ax7.set_xticks([0, 1])
    ax7.set_yticks([0, 1])
    ax7.set_xticklabels(['Normal', 'Anomaly'])
    ax7.set_yticklabels(['Normal', 'Fraud'])

# 8. ROC Curve
ax8 = plt.subplot(3, 3, 8)
if 'Class' in df.columns and 'Combined_Score' in df.columns:
    fpr, tpr, thresholds = roc_curve(df['Class'], df['Combined_Score'])
    roc_auc = auc(fpr, tpr)

    ax8.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (AUC = {roc_auc:.3f})')
    ax8.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--', label='Random')
    ax8.set_xlim([0.0, 1.0])
    ax8.set_ylim([0.0, 1.05])
    ax8.set_xlabel('False Positive Rate')
    ax8.set_ylabel('True Positive Rate')
    ax8.set_title('Receiver Operating Characteristic (ROC) Curve')
    ax8.legend(loc="lower right")
    ax8.grid(True, alpha=0.3)

# 9. Precision-Recall Curve
ax9 = plt.subplot(3, 3, 9)
if 'Class' in df.columns and 'Combined_Score' in df.columns:
    precision_vals, recall_vals, _ = precision_recall_curve(df['Class'], df['Combined_Score'])
    pr_auc = auc(recall_vals, precision_vals)

    ax9.plot(recall_vals, precision_vals, color='blue', lw=2, label=f'PR curve (AUC = {pr_auc:.3f})')
    ax9.set_xlim([0.0, 1.0])
    ax9.set_ylim([0.0, 1.05])
    ax9.set_xlabel('Recall')
    ax9.set_ylabel('Precision')
    ax9.set_title('Precision-Recall Curve')
    ax9.legend(loc="lower left")
    ax9.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('../reports/figures/anomaly_detection_dashboard.png', dpi=300, bbox_inches='tight')
plt.show()

print("Anomaly detection visualizations saved")


# 12. Alert Generation System


print("ALERT GENERATION SYSTEM")
print("=" * 50)

print("\nGenerating actionable alerts...")

# Create alert rules
alerts = []

# Rule 1: Critical risk transactions
critical_transactions = df[df['Risk_Level'] == 'Critical']
if len(critical_transactions) > 0:
    alerts.append({
        'Alert_ID': 'ALT-001',
        'Severity': 'CRITICAL',
        'Description': 'Critical risk transactions detected',
        'Count': len(critical_transactions),
        'Total_Amount': f"${critical_transactions['Amount'].sum():,.2f}",
        'Avg_Risk_Score': f"{critical_transactions['Risk_Score'].mean():.3f}",
        'Action': 'Immediate review and possible block'
    })

# Rule 2: High amount + high risk
high_amount_threshold = df['Amount'].quantile(0.99)
high_risk_high_amount = df[
    (df['Risk_Level'].isin(['Critical', 'High'])) &
    (df['Amount'] > high_amount_threshold)
]
if len(high_risk_high_amount) > 0:
    alerts.append({
        'Alert_ID': 'ALT-002',
        'Severity': 'HIGH',
        'Description': f'High-risk transactions with amount > ${high_amount_threshold:,.2f}',
        'Count': len(high_risk_high_amount),
        'Total_Amount': f"${high_risk_high_amount['Amount'].sum():,.2f}",
        'Action': 'Review within 2 hours, consider additional verification'
    })

# Rule 3: Multiple anomalies from same customer (if CustomerID exists)
if 'CustomerID' in df.columns:
    customer_anomalies = df[df['Final_Anomaly']].groupby('CustomerID').size()
    suspicious_customers = customer_anomalies[customer_anomalies >= 3]

    if len(suspicious_customers) > 0:
        alerts.append({
            'Alert_ID': 'ALT-003',
            'Severity': 'MEDIUM',
            'Description': 'Customers with multiple anomalous transactions',
            'Count': len(suspicious_customers),
            'Customer_Count': len(suspicious_customers),
            'Action': 'Customer account review, possible fraud pattern'
        })

# Rule 4: Unusual V-feature patterns
# Identify transactions with extreme V-feature values
v_anomalies = []
for v_feat in ['V1', 'V2', 'V3', 'V4', 'V7', 'V9', 'V10', 'V11', 'V12', 'V14', 'V16', 'V17', 'V18']:
    if v_feat in df.columns:
        threshold = df[v_feat].abs().quantile(0.995)
        extreme_count = (df[v_feat].abs() > threshold).sum()
        if extreme_count > 0:
            v_anomalies.append({
                'feature': v_feat,
                'count': extreme_count,
                'threshold': threshold
            })

if v_anomalies:
    top_v_anomaly = max(v_anomalies, key=lambda x: x['count'])
    alerts.append({
        'Alert_ID': 'ALT-004',
        'Severity': 'MEDIUM',
        'Description': f"Unusual patterns in {top_v_anomaly['feature']} feature",
        'Count': top_v_anomaly['count'],
        'Threshold': f"{top_v_anomaly['threshold']:.2f}",
        'Action': 'Review feature patterns, check for systemic issues'
    })

# Create alerts dataframe
if alerts:
    alerts_df = pd.DataFrame(alerts)
    print(f"\nGenerated {len(alerts)} alerts")
    print("\nActive Alerts:")
    display(alerts_df[['Alert_ID', 'Severity', 'Description', 'Count', 'Action']])

    # Save alerts
    alerts_df.to_csv('../reports/anomaly_detection/active_alerts.csv', index=False, encoding='utf-8')
    print("\nAlerts saved to: ../reports/anomaly_detection/active_alerts.csv")
else:
    print("No alerts generated - system operating within normal parameters")

# 13. Model Configuration & Deployment

print("MODEL CONFIGURATION & DEPLOYMENT")
print("=" * 50)

print("\nCreating model configuration...")

# Create comprehensive configuration
model_config = {
    'dataset_info': {
        'total_transactions': len(df),
        'fraud_count': int(df['Class'].sum()) if 'Class' in df.columns else 'unknown',
        'fraud_rate': float(df['Class'].mean()) if 'Class' in df.columns else 'unknown',
        'features_used': len(selected_features)
    },
    'detection_methods': {
        'z_score': {
            'optimal_threshold': float(optimal_threshold),
            'features_used': stat_features,
            'performance': zscore_metrics_df[zscore_metrics_df['Threshold'] == optimal_threshold].to_dict('records')[0]
            if 'zscore_metrics_df' in locals() and not zscore_metrics_df.empty else {}
        },
        'isolation_forest': {
            'optimal_contamination': float(optimal_contamination),
            'n_estimators': 100,
            'features_used': iso_features,
            'performance': iso_metrics_df[iso_metrics_df['Contamination'] == optimal_contamination].to_dict('records')[0]
            if 'iso_metrics_df' in locals() and not iso_metrics_df.empty else {}
        },
        'combined_method': {
            'weights': weights,
            'optimal_threshold': float(optimal_combined_threshold),
            'performance': combined_metrics_df[combined_metrics_df['Threshold'] == optimal_combined_threshold].to_dict('records')[0]
            if 'combined_metrics_df' in locals() and not combined_metrics_df.empty else {}
        }
    },
    'risk_scoring': {
        'weights': {
            'zscore_normalized': 0.3,
            'isolation_score_normalized': 0.4,
            'amount_normalized': 0.2,
            'combined_score': 0.1
        },
        'thresholds': {
            'critical': 0.8,
            'high': 0.6,
            'medium': 0.4,
            'low': 0.2
        }
    },
    'alert_system': {
        'rules': [
            {'id': 'ALT-001', 'condition': 'Risk_Level == "Critical"', 'severity': 'CRITICAL'},
            {'id': 'ALT-002', 'condition': 'Risk_Level in ["Critical", "High"] and Amount > percentile_99', 'severity': 'HIGH'},
            {'id': 'ALT-003', 'condition': 'Customer has >= 3 anomalies', 'severity': 'MEDIUM'},
            {'id': 'ALT-004', 'condition': 'Extreme V-feature values', 'severity': 'MEDIUM'}
        ]
    },
    'performance_metrics': {
        'best_method': 'combined_method',
        'overall_anomaly_rate': float(df['Final_Anomaly'].mean()),
        'total_anomalies_detected': int(df['Final_Anomaly'].sum())
    },
    'deployment_info': {
        'model_files': ['isolation_forest_model.pkl', 'robust_scaler.pkl'],
        'feature_columns': selected_features,
        'thresholds': {
            'combined_score': float(optimal_combined_threshold),
            'zscore': float(optimal_threshold)
        },
        'created_date': datetime.now().strftime('%Y-%m-%d %H:%M:%S'),
        'version': '1.0.0'
    }
}

# Save configuration
config_path = '../reports/anomaly_detection/model_configuration.json'
with open(config_path, 'w', encoding='utf-8') as f:
    json.dump(model_config, f, indent=2)

print(f"Model configuration saved: {config_path}")

# Save final results
final_results = df[[
    'id', 'Amount', 'Class' if 'Class' in df.columns else None,
    'ZScore', 'Isolation_Score', 'Combined_Score',
    'Risk_Score', 'Risk_Level', 'Final_Anomaly'
]].dropna(axis=1, how='all')

final_results.to_csv('../reports/anomaly_detection/final_anomaly_results.csv', index=False, encoding='utf-8')
print(f"Final results saved: {len(final_results):,} records")

# %% [markdown]
# ## 14. Final Report & Recommendations

# %%
print("GENERATING FINAL RISK ANALYSIS REPORT")
print("=" * 50)

# Generate comprehensive report
report_date = datetime.now().strftime('%Y-%m-%d %H:%M:%S')
total_transactions = len(df)
total_anomalies = df['Final_Anomaly'].sum()
anomaly_rate = total_anomalies / total_transactions * 100

if 'Class' in df.columns:
    actual_fraud = df['Class'].sum()
    detected_fraud = ((df['Final_Anomaly'] == True) & (df['Class'] == 1)).sum()
    detection_rate = detected_fraud / actual_fraud * 100 if actual_fraud > 0 else 0
    false_positives = ((df['Final_Anomaly'] == True) & (df['Class'] == 0)).sum()
    false_positive_rate = false_positives / (df['Class'] == 0).sum() * 100
else:
    actual_fraud = 'N/A'
    detected_fraud = 'N/A'
    detection_rate = 'N/A'
    false_positives = 'N/A'
    false_positive_rate = 'N/A'

# Calculate total high-risk amount safely
if 'high_risk_transactions' in locals() and len(high_risk_transactions) > 0:
    total_high_risk_amount = high_risk_transactions['Amount'].sum()
else:
    total_high_risk_amount = 0

# Create report
risk_report = f"""
BANK TRANSACTION RISK ANALYSIS REPORT
=====================================
Generated: {report_date}
Total Transactions Analyzed: {total_transactions:,}

EXECUTIVE SUMMARY
=================

1. RISK OVERVIEW
----------------
• Anomalies Detected: {total_anomalies:,} ({anomaly_rate:.4f}%)
• Actual Fraud Cases: {actual_fraud}
• Fraud Detection Rate: {detection_rate if isinstance(detection_rate, str) else f'{detection_rate:.2f}%'}
• False Positive Rate: {false_positive_rate if isinstance(false_positive_rate, str) else f'{false_positive_rate:.4f}%'}

2. DETECTION PERFORMANCE
------------------------
• Best Detection Method: Combined Approach
• Optimal Combined Threshold: {optimal_combined_threshold:.3f}
• Features Used: {len(selected_features)} most correlated features

3. RISK DISTRIBUTION
--------------------
• Critical Risk Transactions: {(df['Risk_Level'] == 'Critical').sum():,}
• High Risk Transactions: {(df['Risk_Level'] == 'High').sum():,}
• Total High-Risk Amount: ${total_high_risk_amount:,.2f}

4. KEY FINDINGS
---------------
• Top predictive feature: {feature_importance.iloc[0]['Feature'] if 'feature_importance' in locals() and len(feature_importance) > 0 else 'N/A'}
• Most effective method: {performance_df.loc[performance_df['F1_Score'].idxmax(), 'Method'] if 'performance_df' in locals() and len(performance_df) > 0 else 'Combined'}
• Alert threshold: {optimal_combined_threshold:.3f} (balances precision and recall)

ACTION REQUIRED
===============

IMMEDIATE ACTIONS (Next 24 hours)
---------------------------------
1. **Review Critical Alerts**
   • Investigate {(df['Risk_Level'] == 'Critical').sum():,} critical risk transactions
   • Focus on transactions > ${high_amount_threshold:,.2f} with high risk scores
   • Implement additional verification for suspicious patterns

2. **System Monitoring**
   • Monitor false positive rate (target: < 5%)
   • Track detection rate (target: > 85%)
   • Review alert effectiveness daily

STRATEGIC RECOMMENDATIONS
=========================

SHORT-TERM (Next 30 days)
-------------------------
1. **System Integration**
   • Implement real-time scoring API
   • Integrate with transaction processing system
   • Develop monitoring dashboard

2. **Process Optimization**
   • Establish investigation workflow
   • Create risk-based approval processes
   • Implement feedback loop for model improvement

LONG-TERM (Next 90 days)
------------------------
1. **Model Enhancement**
   • Implement ensemble methods
   • Add temporal pattern recognition
   • Incorporate customer behavior analysis

2. **Advanced Features**
   • Real-time adaptive thresholds
   • Network analysis for fraud rings
   • Predictive risk scoring

TECHNICAL IMPLEMENTATION
------------------------
1. **Production Deployment**
   • API endpoint for real-time scoring
   • Batch processing for historical analysis
   • Automated model retraining pipeline

2. **Monitoring & Alerting**
   • Real-time dashboard
   • Automated alert escalation
   • Performance tracking

3. **Data Pipeline**
   • Automated feature engineering
   • Quality checks and validation
   • Audit trail for all decisions

PERFORMANCE METRICS
-------------------
• Precision: Target > 80%
• Recall: Target > 85%
• False Positive Rate: Target < 5%
• Average Investigation Time: Target < 2 hours
• Model Retraining: Weekly updates

MODEL FILES GENERATED
---------------------
1. isolation_forest_model.pkl
2. robust_scaler.pkl
3. model_configuration.json
4. feature_importance.csv

ALERT SUMMARY
-------------
Total Active Alerts: {len(alerts) if 'alerts' in locals() else 0}
Critical Alerts: {len([a for a in alerts if a['Severity'] == 'CRITICAL']) if 'alerts' in locals() else 0}
High Priority Alerts: {len([a for a in alerts if a['Severity'] == 'HIGH']) if 'alerts' in locals() else 0}

=====================================
END OF REPORT
"""

# Save report
report_path = '../reports/anomaly_detection/final_risk_report.txt'
with open(report_path, 'w', encoding='utf-8') as f:
    f.write(risk_report)

print("Final risk report generated")
print(f"Report saved to: {report_path}")

# Display summary
print("\nReport Summary:")
print("-" * 40)
print(f"• Total transactions analyzed: {total_transactions:,}")
print(f"• Anomalies detected: {total_anomalies:,} ({anomaly_rate:.4f}%)")
print(f"• Risk levels categorized: {len(risk_analysis)}")
print(f"• Model performance: Combined method with threshold {optimal_combined_threshold:.3f}")
print(f"• Files generated: See ../reports/anomaly_detection/")

# %% [markdown]
# ## 15. Summary

# %%
print("ANOMALY DETECTION ANALYSIS COMPLETE!")
print("=" * 50)
print("\nGenerated Files:")
print("   ../reports/anomaly_detection/")
print("     - fraud_distribution.csv")
print("     - feature_correlations.csv")
print("     - selected_features.csv")
print("     - zscore_metrics.csv")
print("     - isolation_forest_metrics.csv")
print("     - combined_metrics.csv")
print("     - performance_comparison.csv")
print("     - classification_report.csv")
print("     - risk_level_distribution.csv")
print("     - risk_level_financials.csv")
print("     - high_risk_transactions.csv")
print("     - feature_importance.csv")
print("     - active_alerts.csv")
print("     - final_anomaly_results.csv")
print("     - model_configuration.json")
print("     - final_risk_report.txt")
print("\n   ../reports/figures/")
print("     - anomaly_detection_dashboard.png")
print("\n   ../models/")
print("     - isolation_forest_model.pkl")
print("     - robust_scaler.pkl")
print("\nKey Results:")
print(f"   • Total anomalies detected: {total_anomalies:,}")
print(f"   • Anomaly detection rate: {anomaly_rate:.4f}%")
print(f"   • High-risk transactions: {len(high_risk_transactions) if 'high_risk_transactions' in locals() else 0:,}")
print(f"   • Active alerts generated: {len(alerts) if 'alerts' in locals() else 0}")
print(f"   • Optimal combined threshold: {optimal_combined_threshold:.3f}")
print("\nPROJECT COMPLETED SUCCESSFULLY!")
```
